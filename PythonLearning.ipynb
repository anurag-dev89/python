```python
#set
p1 = {"python", "sql", "ml" ,"excel"}
print(p1)
p2 = {"python", "sql", "ai", "ml", "c++"}
#find common subjects
common = p1 & p2
print(common)

#find subjects in p1 but not in p2
subjects_in_p1_not_in_p2 = p1 - p2
print(subjects_in_p1_not_in_p2)

#symmetric difference
symmetric_diff = p1 ^ p2
print(symmetric_diff)
```
