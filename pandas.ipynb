#10 concatenation and merge dataframes

```python
df1 = pd.DataFrame({'ID': [1,2,3], 'name':['A','B','C']})
df1
```

```python
df2 = pd.DataFrame({'ID': [1,2,2,4], 'score':[88, 96, 78, 90]})
df2
```

```python
pd.concat([df1, df2], axis=0) #concatenate the two dataframes vertically
```

```python
pd.concat([df1, df2], axis=1) #concatenate the two dataframes horizontally
```

```python
#merge
pd.merge(df1, df2, on='ID', how='inner') #inner join
```

```python
#pd.merge(df1, df2, on='ID', how='right') #right join
#pd.merge(df1, df2, on='ID', how='left') #left join
pd.merge(df1, df2, on='ID', how='outer') #outer join
```
