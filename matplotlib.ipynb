```python
#import matplotlib
import matplotlib.pyplot as plt
#Pyplot is a module of Matplotlib that provides functions to create, customize, and display graphs and charts.
```

#1 create first chart

```python
x=[1,2,3,4,5,]
y=[10,20,25,35,45]

plt.plot(x,y) #line chart
plt.show() #displays graph in a separate window or inline in a Jupyter Notebook.
```

#2 customizing charts

```python
plt.figure(figsize=(5,3)) #size of the figure in inches (width, height)
plt.plot(x ,y , color = 'red' ,marker ='o', linestyle ='dashed',linewidth =2 ,markersize =8) #customizing the line chart

plt.title('Line Chart') #title of the chart
plt.xlabel('X-axis') #label for x-axis
plt.ylabel('Y-axis') #label for y-axis

plt.show()
```

#3 advanced line chart - multiple line chart & legends

```python
#A line chart uses points connected by lines to show changes or trends over time or across an ordered sequence.
x=[1,2,3,4,5,]
y1=[10,20,25,35,45]
y2=[20,30,35,45,55]

plt.plot(x,y1, label = 'sales 2024')
plt.plot(x,y2, label = 'sales 2025')

plt.title('line chart')
plt.xlabel('months')
plt.ylabel('sales')

plt.legend() #legend displays the labels for each line chart
plt.show()
```

#4 Bar chart

```python
#A bar chart uses rectangular bars to compare values across different categories.
x=['A','B','C','D','E',]
y=[10,20,55,35,45]

plt.bar(x,y) #bar chart

plt.title('Bar Chart') #title of the chart
plt.show()
```

#5 Histogram

```python
#A histogram shows the frequency distribution of numerical data by grouping values into intervals called bins.
import random as rd

data = rd.choices(range(1, 101), k=100)
plt.hist(data , bins =8)
#plt.hist(data , bins =8 , color = 'green' , edgecolor = 'black') #customizing the histogram
plt.title('Histogram') #title of the chart
plt.show()
```

#pie chart

```python
#A pie chart represents data as slices of a circle. Each slice shows a category's proportion of the total.
categories=['A','B','C','D','E',]
sales=[10,20,55,35,45]

plt.pie(sales , labels = categories , autopct = '%1.1f%%') #autopct displays the percentage of each slice
plt.title('Pie Chart') #title of the chart
plt.show()
```

#7 scatter plot

```python
#A scatter plot uses individual dots to show the relationship between two numerical variables.
study_hr = [1, 2, 3, 4, 5, 6]
marks = [35, 40, 50, 55, 70, 85]

plt.scatter(study_hr ,marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

#8 subplots

```python
#A subplot allows you to display multiple charts inside a single figure.

#data 1 - bar chart
categories=['Mon','Tue','Wed','Thu','Fri',]
sales=[10,20,55,35,45]

#data 2 -scatter plot
x=['A','B','C','D','E',]
y=[10,20,55,35,45]

plt.figure(figsize=(6,4))
plt.subplot(1,2,1) #1 row, 2 columns, 1st subplot

#first subplot - bar chart
plt.bar(categories , sales)
plt.xlabel('Days of the week')
plt.ylabel('sales')
plt.title('Bar Chart')

#second subplot - scatter plot
plt.subplot(1,2,2) #1 row, 2 columns, 2nd subplot
plt.scatter(x , y)
plt.title('Scatter Plot')

plt.show()
```

#9 matplotlib with pandas -real data

```python
#create df
import pandas as pd

#data
data ={
    'months' : ['jan','feb','mar','apr','may'],
    'sales' : [1000 , 2500 , 2000 , 4500 , 3000]
}

df = pd.DataFrame(data)
df
```

```python
plt.bar(df['months'] ,df['sales'])
plt.title('Sales over months')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.show()
```

#10 save plots/charts

```python
plt.bar(df['months'] ,df['sales'])
plt.title('Sales over months')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.savefig('sales_chart.png') #saves the chart as a PNG file

plt.show()
```
